```python
x=int(input("enter x:"))
y=int(input("enter y:"))
z=int(input("enter z:"))

def cal_sum(x,y,z):
    sum=x+y+z
    return sum
s=cal_sum(x,y,z)
print(s)
```

```python
def cal_sum_prod(x,y,z):
    ss=x+y+z
    pp=x*y*z
    return ss,pp
a=int(input("enter a:"))
b=int(input("enter b:"))
c=int(input("enter c:"))
s,p=cal_sum_prod(a,b,c)
print(s,p)

```

```python
def prime_factor():
    n=int(input("enter a number:"))

    for i in range(2,n+1):
        if n%i == 0:
            print(i, end=" ")
            while n%i ==0:
                n=n//i
prime_factor()
```

```python
def even_numbers():
    for i in range(1,101):
        if i%2==0:
            print(i,end=" ")
even_numbers()
```
