```python
number=[20,30,40,50]
print("the 1st element is:",number[0])
print("the 2nd number is:",number[1])
print("the 3rd number is:",number[2])
print("the 4th number is:",number[3])
```

```python
#to append:which is to add no to the end of the list
number.append(60)
print("\nlist after appending:",number)
```

```python
#to reverse the list 
number.reverse()
print("\nlist after reverse:",number)
```

```python
#to print the occurences of the element in the cell
numbers=[10,20,30,30,40,50]
element=int(input("Enter the element:"))
count=numbers.count(element)
print("number of occurence are:",count)
```

```python
#to append the items of list1 to list2 in the front
list1=[20,30,40]
list2=[50,60,90]
list2=list1+list2
print("list1:",list1)
print("list2:",list2)
```

```python
#to insert new element to any index of the list
numbers=[10,20,30,40,50]
numbers.insert(0,90)
print("the new list is:",numbers)
```

```python
#to remove an element on the basis of the specific index
numbers=[10,20,30,40,50]
index=int(input("enter the index of the element to be removed:"))
numbers.pop(index)
print("after removal list:",numbers)
```
