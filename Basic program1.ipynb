```python
#basic calculator
num1=float(input("Enter the first no:"))
num2=float(input("Enter the second no:"))
print("\nchoose an operator:,1.add,2.substract,3.divide,4.multiply")
choice=int(input("Enter your choice(1-4):"))
if choice==1:
    print("\nthe addition of the numbers provided is:",num1+num2)
elif choice==2:
    print("\nthe substraction of the numbers provide is:",num1-num2)
elif choice==3:
    print("\nthe division of the numbers provided is:",num1/num2)
else:
    print("\nthe multiplication of the numbers provide is:",num1*num2)
```

```python
#calculate percentage and display the highest and lowest marks aslo accept the student details

name=str(input("Enter your name:"))
rollno=int(input("Enter your roll no:"))
#accept marks for 3 subjects

subject1=input("enter your subject1 name:")
marks1=float(input("enter marks for subject1:"))

subject2=input("enter your subject2 name:")
marks2=float(input("enter marks for subject2:"))

subject3=input("enter your subject3 name:")
marks3=float(input("enter marks for subject3:"))

#percentage
total=marks1+marks2+marks3
percentage=(total/300)*100

#show student details
print("\n___STUDENT_RESULT___")
print("Student name:",name)
print("Roll number:",rollno)
print("Student percentage:",percentage)

#show highest and lowest marks
if marks1>=marks2 and marks1>=marks3:
    print("Subject with highest marks is :",subject1)
    if marks1==marks2:
        print("subjects with highest marks are:",subject1,subject2)
    else:
        print("subjects with highest marks are:",subject1,subject2)
    
elif marks2>=marks1 and marks2>=marks3:
    print("Subject with highest marks is :",subject2)
    if marks2==marks1:
        print("subjects with highest marks are:",subject1,subject2)
    else:
        print("subjects with highest marks are:",subject2,subject3)
     
else:
     print("Subject with highest marks is :",subject3)
     if marks3==marks1:
         print("subjects with highest marks are:",subject1,subject3)
     else:
         print("subjects with highest marks are:",subject3,subject2)
         
if marks1<=marks2 and marks1<=marks3:
    print("Subject with lowest marks is :",subject1)
  
elif marks2<=marks1 and marks2<=marks3:
     print("Subject with lowest marks is :",subject2)
    
else:
     print("Subject with lowest marks is :",subject3)
   
```

```python
#to reverse the digits of a number
num=int(input("enter a number:"))
reverse=0
while num>0:
    digit=num%10
    reverse=reverse*10+digit
    num=num//10
print("reverse:",reverse)
```
