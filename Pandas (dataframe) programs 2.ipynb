```python
#DataFrames
##data sets in pandas are usually multi-dimensional tables, called data frames. Series is like a column, a dataframe is a whole table
import pandas as pd
data={
    "calories": [420,300,390],
    "duration": [50,40,45]
}
myvar = pd.DataFrame(data)
print(myvar)
```

```python
#locate row:as you can see the result above,the dataframe is like a tabole with rows and columns.
##Pandas use the loc attribute to return one or more specified rows
import pandas as pd
data={
    "calories": [420,300,390],
    "duration": [50,40,45]
}
#load data into a dataframe object:
df=pd.DataFrame(data)
print(df.loc[0])
```

```python
#Named indexes:with the index argument, you can name your own indexes
import pandas as pd
data={
    "calories": [420,300,390],
    "duration": [50,40,45]
}
#load data into a dataframe object:
df=pd.DataFrame(data, index=["day1","day2","day3"])
print(df)
```

```python
#data cleaning and preprocessing using pandas:its means fixing and organising messy data.it has tools and funtions to help us clean the data.
#Data cleaning often involves:
##1.Dropping irrelevant columns
##2.Renaming Columns names into meaningful names
##3.Making data values consistent
##4.Replacing or filling in missing values
import pandas as pd
#define adictionary with sample data which includes some missing values
data={
    'A': [1,2,3,None,5],
    'B': [None,2,3,4,5],
    'C': [1,2,None,4,5]
}
df = pd.DataFrame(data)
print("Orignal Data:\n",df)
print()

# use dropna() to remove rows with any missing values
df_cleaned=df.dropna()
print("Cleaned Data:\n",df_cleaned)
```

```python
import pandas as pd
#define adictionary with sample data which includes some missing values
data={
    'A': [1,2,3,None,5],
    'B': [None,2,3,4,5],
    'C': [1,2,None,4,5]
}
df = pd.DataFrame(data)
print("Orignal Data:\n",df)
print()
#Filling NaN values with 0
df.fillna(0,inplace=True)
print("Cleaned Data:\n",df)
```

```python
#use aggregate funtion to fill missing value
import pandas as pd
#define adictionary with sample data which includes some missing values
data={
    'A': [1,2,3,None,5],
    'B': [None,2,3,4,5],
    'C': [1,2,None,None,5]
}
df = pd.DataFrame(data)
print("Orignal Data:\n",df)
print()
#Filling NaN values with 0
df.fillna(df.mean(),inplace=True)
print("Cleaned Data:\n",df)
```

```python
#how to handle duplicate values
import pandas as pd
#sample data
data={
    'A': [1,2,2,3,4],
    'B': [5,6,6,7,8]
}
df=pd.DataFrame(data)
print("Orignal Dataframe:\n", df.to_string(index=False))
#detect duplicates
print("Duplicte Rows=:\n", df[df.duplicated()].to_string(index=False))
#remove duplicates based on column 'A'
df.drop_duplicates(subset=['A'], keep='first', inplace=True)
print("DataFrame after removing duplicates based on column A:\n",df.to_string(index=False))
## df[df.duplicated()] produces a boolean series to identify the duplicate values
## df.drop_duplicates(subset=['A'],keep='first',inplace=True) removes duplicated based on column a retaining
#only the first occurence of each duplicate directely in orignal dataframe
```

```python
#rename column names to meaningful names: we can use the rename() funtion
import pandas as pd
#sample data
data={
    'A': [25,30,35],
    'B':['Richie','Daisy','Smith'],
    'C': [500000,90000000,700000]
}
df=pd.DataFrame(data)
#rename columns
df.rename(columns={'A':'Age','B':'Name','C':'Salary'}, inplace=True)
print(df.to_string(index=False))
```
