```python
#What is pandas?
#pandas, is a series which is a one-dimensional labbeled array capable of holding a datatype
# for panda the short cut is pd
##1 CREATING A SERIES FROM A LIST
import pandas as pd
a = [1,7,2]
myvar =  pd.Series(a)
print(myvar)
```

```python
##2 CREATING LABELS WITH THE INDEX ARGUMENT ,YOU CAN NAME YOUR OWN LABELS
import pandas as pd
a = [1,7,2]
myvar =  pd.Series(a, index=["x","y","z"])
print(myvar)
```

```python
import pandas as pd
s1 = pd.Series([10,20,30], index=['a','b','c'], name='Example Series')
print(s1)
```

```python
##3 CREATING A SERIES FROM A DICTIONARY
import pandas as pd
s2 = pd.Series({'X':1, 'Y':2, 'Z':3})
print(s2)
```

```python
import pandas as pd
calories = {'day1':420,'day2':390,'day3':380}
myvar =  pd.Series(calories)
print(myvar)
```
