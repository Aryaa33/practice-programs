```python
##2 keyword argument
#idea is to allow the caller to specify the argument name with values so that the caller does not 
#need to remmember the order of the parameters
def student(fname,lname):
    print(fname,lname)
student(fname="good",lname="practice")
student(lname="practice",fname="good")
```

```python
##3 positional argument
def nameAge(name,age):
    print("hi,I am ",name)
    print("My age is ",age)
print("case-1:")
nameAge("suraj",27)
print("\ncase-2")
nameAge(27,"suraj")
```

```python
#arbitrary keyword argument
#used to define anything in general
#argument with single*args for arbitrary positional argument.
#argument with double**kwargs for arbitrary positional argument.
def myFun(*argv):
    for arg in argv:
        print(arg)
myFun('hello','welcome','to','MIT')
```

```python
#Recursive function
#function that repeats itself

def factorial(n):
    if n==0:
        return 1
    else:
        return n*factorial(n-1)
print(factorial(4))
```
