```python
#an array that has 1-d arrray as its elements is called as 2-d array. they often represent a matrix
#2-d arrays containing 2 arrays
import numpy as np
arr=np.array([[1,2,3],[4,5,6]])
print(arr)
```

```python
#3-d array with 2 2-d arrays in it
import numpy as np
arr=np.array([[[1,2,3],[4,5,6]],[[1,2,3],[4,5,6]]])
print(arr)
```

```python
#numpy arrays provides the ndim attribute tht returns and integer tht tells us how many dimensions the arrays have
import numpy as np

a = np.array(42)
b = np.array([1,2,3,4,5,6])
c = np.array([[1,2,3],[4,5,6]])
d = np.array([[[1,2,3],[4,5,6]],[[1,2,4],[5,6,7]]])


print(a.ndim)
print(b.ndim)
print(c.ndim)
print(d.ndim)
```
