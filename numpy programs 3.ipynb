```python
#create a 3x3 numpy array of all true
import numpy as np

arr = np.ones((3, 3), dtype=bool)
print(arr)
```

```python
#reverse a 1d numpy array
import numpy as np

arr = np.array([1, 2, 3, 4, 5])
reversed_arr = arr[::-1]

print(reversed_arr)
```

```python
#create a 3x3 identity matrix
import numpy as np

identity = np.eye(3)
print(identity)
```

```python
#convert a python list into numpy list
import numpy as np

my_list = [1, 2, 3, 4, 5]
arr = np.array(my_list)

print(arr)
```

```python
#perform arithmetic operation on two numpy arrays element wise
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(a + b)  # Addition
print(a - b)  # Subtraction
print(a * b)  # Multiplication
print(a / b)  # Division
```
