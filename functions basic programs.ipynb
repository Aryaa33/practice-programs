```python
#introduction to control structure and functions
##conditional statements and loop control statements and funtions - defn and parameters
##a) if statement:executes block of code if its true
##b) if else statement:executes on block of code if its true and another block if its false
##c) if elif else statement: allows checking multiple conditions
score=85
attendance=93
if score >= 70:
    print("student has passed the exam")
    if attendance >= 90:
        print("student also has excellent attendance")
    else:
        print("students attendance could be improved")
else:
    print("student did not pass the exam")
```

```python
#looping statements(iteration):these statement allow a block of vode to be executed repeatedly
## for loop:iterates over a sequence(like a list,tuple,string, or range)
fruits=["apple","banana","cherry"]
for fruits in fruits:
    print(fruits)
```

```python
## while loop executes a block of code as long as the a condition remains true
count=0
while count<5:
    print(count)
    count+=1
```

```python
##nested loop
adj=["red","big","tasty"]
fruits=["apple","banana","cherry"]
for x in adj:
    for y in fruits:
        print(x,y)
```

```python
##breaks statement:terminates the current loop entirely and execution resumes at the statement immediately following the loop
for i in range(10):
    if i==5:
        break
    print(i)
```
