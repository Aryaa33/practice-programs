```python
#1 to add to dictionary
my_dict={0:10,1:20}
my_dict[2]=30
print("Dictionary:",my_dict)
```

```python
#2 to make 3 dictionaries to 1 dictionary
dic1={1:20,2:20}
dic2={3:30,4:40}
dic3={5:50,6:60}
new_dict={}
new_dict.update(dic1)
new_dict.update(dic2)
new_dict.update(dic3)
print("new dictionary:",new_dict)
```

```python
#3 to check if key exists in dictionary
my_dict={101:"vedanshi",102:"shardul",103:"aditya",104:"jonny"}
key=int(input("enter key to search:"))
if key in my_dict:
    print("key exists in the dictionary")
else:
    print("key does not exist in the dictionary")
```

```python
#4 to get the key and its value
my_dict={
    101:"vedanshi",
    102:"aditya",
    103:"shardul",
    104:"jonny",
}
print("keys:")
for key in my_dict:
    print(key)
print("\nValues:")
for values in my_dict.values():
    print(values)
print("\nKeys and value")
for key, value in my_dict.items():
    print(key,value)
```

```python
#5 to sq all number included in the given range
my_dict={}
for i in range(1,16):
    my_dict[i]=i**2
print(my_dict)
```

```python
#6 program to sum all values in dictionary
my_dict={"a":10,"b":20,"c":30,"d":40}
total=0
for value in my_dict.values():
    total=total+value
print("sum of value:",total)
```
